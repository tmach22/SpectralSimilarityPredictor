## Analyze the GNPS cleaned dataset

```python
from matchms.importing import load_from_mgf
```

```python
# SAFE: This creates a "stream" and uses almost zero RAM initially
spectra_generator = load_from_mgf("/data/nas-gpu/wang/tmach007/SpectralSimilarityPredictor/spectra_pairs/ALL_GNPS_cleaned.mgf")
```

```python
# Process spectra one-by-one
for spectrum in spectra_generator:
    # Assuming you have already run: first_item = next(spectra_generator)

    print("="*30)
    print("  SPECTRUM OBJECT SUMMARY  ")
    print("="*30)

    # 1. Print all Metadata (Headers)
    print("\n--- [METADATA] ---")
    for key, value in spectrum.metadata.items():
        print(f"{key:20}: {value}")

    # 2. Print Peak Information (The Data)
    print("\n--- [PEAK DATA] ---")
    mz = spectrum.peaks.mz
    intensities = spectrum.peaks.intensities

    print(f"Total Number of Peaks: {len(mz)}")
    print(f"Mass Range (m/z):    {mz.min():.4f} - {mz.max():.4f}")
    print(f"Max Intensity:       {intensities.max():.2f}")

    # 3. Print the first 10 peaks to see the format
    print("\n--- [FIRST 10 PEAKS (m/z, intensity)] ---")
    for m, i in zip(mz[:10], intensities[:10]):
        print(f"{m:10.4f} \t {i:10.2f}")

    # 4. Print Calculated Attributes (if any)
    if spectrum.losses:
        print(f"\nNeutral Losses detected: {len(spectrum.losses.mz)} peaks")
```
