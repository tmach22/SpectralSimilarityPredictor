## Analyze the GNPS cleaned dataset

```python
import os
import pickle
from multiprocessing import Pool, cpu_count
from matchms.importing import load_from_mgf
from matchms.exporting import save_as_mgf
```

```python
def process_and_save_chunk(chunk_data):
    """Worker function to process and save a single chunk."""
    spectra, chunk_id, output_dir = chunk_data
    
    # 1. Apply your existing preprocessing here
    # (e.g., select_by_mz, normalize_intensities, etc.)
    processed_spectra = spectra  # Placeholder for your cleaning logic
    
    # 2. Save the chunk as an MGF file
    chunk_filename = os.path.join(output_dir, f"chunk_{chunk_id}.mgf")
    save_as_mgf(processed_spectra, chunk_filename)
    
    return f"Successfully saved {chunk_filename}"
```

```python
def run_parallel_pipeline(input_file, output_dir, chunk_size=10000):
    os.makedirs(output_dir, exist_ok=True)
    
    # Use a generator to read without loading 1.8GB into RAM
    spectra_generator = load_from_mgf(input_file)
    
    chunk_id = 0
    current_chunk = []
    
    # Initialize Multiprocessing Pool
    with Pool(processes=cpu_count()) as pool:
        tasks = []
        
        for spectrum in spectra_generator:
            current_chunk.append(spectrum)
            
            # When chunk is full, prepare it for parallel processing
            if len(current_chunk) >= chunk_size:
                tasks.append((current_chunk, chunk_id, output_dir))
                current_chunk = []
                chunk_id += 1
                
                # Execute tasks in parallel when we have enough batches
                if len(tasks) >= cpu_count():
                    for res in pool.map(process_and_save_chunk, tasks):
                        print(res)
                    tasks = []
        
        # Process any remaining spectra in the final partial chunk
        if current_chunk:
            tasks.append((current_chunk, chunk_id, output_dir))
            for res in pool.map(process_and_save_chunk, tasks):
                print(res)
```
